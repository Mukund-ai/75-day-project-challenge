```python
#  create a dirty dataset
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 104, 105, None],
    "Name": ["Amit", "Sneha", "Rahul", "Priya", "Priya", "Rohit", "Meera"],
    "Age": [25, None, 35, 29, 29, 40, 22],
    "City": ["Delhi", "Mumbai", None, "Delhi", "Delhi", "Chennai", "Chennai"],
    "Purchase_Amount": [1200, 1500, None, 1800, 1800, 2200, 900]
}

df = pd.DataFrame(data)
df
```

```python
#data overview
df.info()
```

```python
#Find any missing data in ther or not
df.isnull().sum()
```

```python
#Remove duplicates data
df = df.drop_duplicates()
df

```

```python
#Handle missing customer Ids
df = df.dropna(subset=["Customer_ID"])
df
```

```python
df["City"].fillna("Unknown", inplace=True)
df


```
